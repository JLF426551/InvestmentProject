**Data Preparation**

`PERCENT` column has values as 'xx.xxx%' which is reading as a string <br/>
Need to update values to read as purely decimal. e.g. 11.0% will become 0.11, 2.4% to 0.0024.


```python
# Correct format
def outputAsDecimal( percentString):
    #Drops the '%' which is always last.
    stringNoPercent = percentString[:-1]
    decimNumber = float( stringNoPercent )
    return decimNumber
```

```python
contrafundDF['PCT'] = contrafundDF['PERCENT'].apply( outputAsDecimal )
divIntlFundDF['PCT'] = divIntlFundDF['PERCENT'].apply( outputAsDecimal )
usSusFundDF['PCT'] = usSusFundDF['PERCENT'].apply( outputAsDecimal )
nasdaqCompDF['PCT'] = nasdaqCompDF['PERCENT'].apply( outputAsDecimal )
intlSusFundDF['PCT'] = intlSusFundDF['PERCENT'].apply( outputAsDecimal )
zeroIntlDF['PCT'] = zeroIntlDF['PERCENT'].apply( outputAsDecimal )
zeroMktDF['PCT'] = zeroMktDF['PERCENT'].apply( outputAsDecimal )
```

**Task 1**: Analyze number of holdings based on percent held. Output number of holdings at: top 20%; top33%; top 50%; and, percent for top ten holdings.

```python
def getTopHeldInformation(mPercentSeries):
    mTop25 = 0
    mTop33 = 0
    mTop50 = 0
    mTen = 0.00
    runningTotal = 0.00

    counter = 0
    hasNotCompleted = True

    #Loop only while we haven't calculated all values.
    #Loop does not consider reaching end of mPercentSeries based on context.
    while hasNotCompleted:
        runningTotal = runningTotal + mPercentSeries.iloc[counter]
        counter = counter + 1
        
        if runningTotal >= 50:
            mTop50 = counter
        elif runningTotal >= 33:
            mTop33 = counter
        elif runningTotal >=25:
            mTop25 = counter

        if counter == 10:
            mTen = runningTotal
            #print("The top 10 reached")

        #Check for conditions to finish loop.
        if runningTotal >= 50 and counter >= 10:
            hasNotCompleted = False
    
    #Return all values as list
    return [ mTop25, mTop33, mTop50, mTen ]

```

```python
contraList = getTopHeldInformation(contrafundDF['PCT'])
contraList = summaryList[0] + contraList
diversList = getTopHeldInformation(divIntlFundDF['PCT'])
diversList = summaryList[1] + diversList 
usSustList = getTopHeldInformation(usSusFundDF['PCT'])
usSustList = summaryList[2] + usSustList
nasdaqList = getTopHeldInformation(nasdaqCompDF['PCT'])
nasdaqList = summaryList[3] + nasdaqList
intSusList = getTopHeldInformation(intlSusFundDF['PCT'])
intSusList = summaryList[4] + intSusList
zeroInList = getTopHeldInformation(zeroIntlDF['PCT'])
zeroInList = summaryList[5] + zeroInList
zeroUSList = getTopHeldInformation(zeroMktDF['PCT'])
zeroUSList = summaryList[6] + zeroUSList
```

```python
def calculateOverlap( seriesA, seriesB):
    serALen = len(seriesA)
    serBLen = len(seriesB)
    tLen = min( serALen, serBLen)
    runningOverlap = 0.000

    for i in range( tLen):
        runningOverlap = runningOverlap + min( seriesA[i], seriesB[i] )

    return runningOverlap
```

```python
#Run international analysis
#Example of hard-coded column selection
intlOverlap = []
intlOverlap.append( calculateOverlap( intlDF['FDIVX'], intlDF['FNIDX']) )
intlOverlap.append( calculateOverlap( intlDF['FDIVX'], intlDF['FZILX']) )
intlOverlap.append( calculateOverlap( intlDF['FNIDX'], intlDF['FZILX']) )
```

```python
#Run domestic overlap analysis
#Exacmple of automated columns selection

#Extract columns.
#List has mutual funds tickers and the 'TICKER' column, will need to drop that.
usDF = usDF.drop( columns = 'TICKER')

mCols = usDF.columns
print(mCols)

usOverlapVal = []
usOverlapTXR = []

#Run nested loop to run comparisons
for i in range( len(mCols) - 1 ):
    for s in range( i+1, len(mCols)):
        print("i: ", i, "s: ",s)
        usOverlapVal.append( calculateOverlap( usDF[mCols[i]], usDF[mCols[s]]) )
        usOverlapTXR.append( [ mCols[i], mCols[s] ])
```
